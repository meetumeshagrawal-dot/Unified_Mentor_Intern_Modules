- Create NumPy vectors and inspect their shape and data type.
- Select elements and slices using zero-based indexing.

```python
import numpy as np

np.set_printoptions(precision=2, suppress=True)
```

## 1. Creating vectors

A NumPy vector is a one-dimensional `ndarray`. Unlike a regular Python list, a NumPy array supports efficient numerical operations.

```python
list_values = [2, 4, 6, 8]
v = np.array(list_values)

print("Vector:", v)
print("Shape:", v.shape)
print("Number of dimensions:", v.ndim)
print("Data type:", v.dtype)
```

```python
# Useful ways to create vectors
zeros = np.zeros(4)
ones = np.ones(3)
sequence = np.arange(1, 6)       # 1 through 5; the stop value is excluded
evenly_spaced = np.linspace(0, 1, 5)  # 5 evenly spaced values, including endpoints

print("Zeros:", zeros)
print("Ones:", ones)
print("Sequence:", sequence)
print("Evenly spaced:", evenly_spaced)
```

## 2. Indexing and slicing

Indexing starts at `0`. A slice `start:stop` includes `start` and excludes `stop`. Negative indices count from the end.

```python
scores = np.array([72, 85, 91, 68, 95])

print("First score:", scores[0])
print("Last score:", scores[-1])
print("Middle scores:", scores[1:4])
print("Scores at least 90:", scores[scores >= 90])
```

## 3. Vector arithmetic

Arithmetic with a scalar is applied to every element. Arithmetic between equal-shaped vectors is element-wise. This is different from the dot product, introduced below.

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print("a + b:", a + b)
print("b - a:", b - a)
print("a * b (element-wise):", a * b)
print("a squared:", a ** 2)
print("a scaled by 3:", 3 * a)
```

## 4. Reductions and the dot product

`sum`, `mean`, `min`, and `max` summarize a vector. The dot product multiplies corresponding entries and adds those products; the vectors must have the same length.

```python
x = np.array([3, 4])
y = np.array([2, 5])

print("Sum:", x.sum())
print("Mean:", x.mean())
print("Minimum and maximum:", x.min(), x.max())
print("Dot product:", np.dot(x, y))  # 3*2 + 4*5 = 26
print("Euclidean length of x:", np.linalg.norm(x))  # sqrt(3**2 + 4**2) = 5
```

## 5. Solved application: convert temperatures

Convert Celsius values to Fahrenheit using the vectorized formula `F = (9/5)C + 32`.

```python
celsius = np.array([0, 20, 37, 100])
fahrenheit = (9 / 5) * celsius + 32

print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
assert np.allclose(fahrenheit, [32, 68, 98.6, 212])
```

Try each prompt before reading its solution cell. The assertions check the requested results.

### Problem 1

Create a vector containing the integers from 5 through 10. Calculate its sum and mean.

```python
values = np.arange(5, 11)
values_sum = values.sum()
values_mean = values.mean()

print("Values:", values)
print("Sum:", values_sum)
print("Mean:", values_mean)
assert values_sum == 45 and values_mean == 7.5
```

### Problem 2

Given `prices` and `quantities`, calculate each item's cost and the total cost. Use element-wise multiplication.

```python
prices = np.array([2.50, 1.25, 4.00])
quantities = np.array([4, 3, 2])
item_costs = prices * quantities
total_cost = item_costs.sum()

print("Cost per item:", item_costs)
print("Total cost:", total_cost)
assert np.allclose(item_costs, [10.0, 3.75, 8.0])
assert np.isclose(total_cost, 21.75)
```

### Problem 3

For the vector below, select only values greater than the mean and count how many values meet that condition.

```python
measurements = np.array([4, 8, 6, 10, 2])
average = measurements.mean()
above_average = measurements[measurements > average]

print("Mean:", average)
print("Above average:", above_average)
print("Count above average:", above_average.size)
assert np.array_equal(above_average, [8, 10])
```

## Key takeaways

- Use `np.array` to make a vector; inspect its `shape`, `ndim`, and `dtype`.
- NumPy uses zero-based indexing and end-exclusive slices.
- Vector expressions apply operations to all elements without an explicit Python loop.
- Use reductions such as `.sum()` and `.mean()` for summaries, and `np.dot` or `@` for a dot product.
- Check array shapes before combining vectors.
